```python
import matplotlib.pyplot as plt
```

```python
student = ['suraj', 'kamal', 'rahul', 'mohan', 'Deepak']
marks = [60, 75, 85, 50, 90]
plt.figure(figsize=(10, 5), facecolor='#121212')
plt.gca().set_facecolor('#1e1e1e')
plt.bar(student, marks, color=['#ff4d4d','#4da6ff','#66ff66','#ffcc66','#ff66cc'])
plt.xlabel('Students naam', color='red')
plt.ylabel('Students Marks', color='red')
plt.title('Students ke marks compare ho rahe hain', color='red')

plt.grid(color='gray', linestyle='--', linewidth=0.5)

# Tick color (important 🔥)
plt.xticks(color='white')
plt.yticks(color='white')

# Value labels  
for i in range(len(student)):
    label = student[i] + " - " + str(marks[i])
    
    plt.text(student[i], marks[i] + 2, label,
             ha='center',
             color='red')

plt.show()
```

```python
X = ['A', 'B', 'C', 'D', 'E']
Y = [10, 20, 30, 40, 50]
plt.figure(figsize=(10, 5), facecolor='black')
plt.bar(X, Y, color=['lightpink', 'BLUE', 'YELLOW', 'GREEN', 'RED'])
plt.title('Student bar chart', color='White', size=20)
plt.xlabel("Char", color='White', size=30)
plt.ylabel("Number", color='White', size=30)
plt.xticks(color='white', size=20)
plt.yticks(color='white', size=20)
plt.grid(color="gray", linestyle='--', linewidth='0.5')
```
